## Diretrizes para o MLP

1.  **Pré-processamento:** É obrigatório aplicar **escalonamento (Padronização ou Normalização)** aos dados de entrada, pois o MLP é sensível à magnitude das *features*.
2.  **Arquitetura:** Construir uma Rede Neural MLP com pelo menos uma camada oculta.
3.  **Configuração de Ativação:**
    * **Camadas Ocultas:** O uso da função **ReLU** nas camadas ocultas é sugerido.
    * **Camada de Saída:** A camada final deve ter **1 neurônio** com uma função de ativação **Sigmoide** (ou ser usada com `BCEWithLogitsLoss`, que a embute).
4.  **Função de Perda:** Utilizar a função de perda **BCELoss** ou **BCEWithLogitsLoss** (Erro de Entropia Cruzada Binária), adequada para classificação binária.
5.  **Avaliação:** Avaliar a performance do modelo usando o conjunto de teste, focando nas métricas de classificação: **Acurácia, Precisão, Recall, F1-Score e Matriz de Confusão**.

## Preparar dados para treinamento da rede neural

```python
# Dividir o dataset em X e y
X = df_parkisons.drop(columns=target, axis=1)
y = df_parkisons[target]
```

```python
# Dividir entre treino e teste
y = y.values
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Dividir entre Validação e Teste Final
X_test, X_val, y_test, y_val = train_test_split(X_test, y_test, test_size=0.5, random_state=42)
```

```python
# Criar transformador de coluna
numeric_transformer = StandardScaler()

preprocessor = ColumnTransformer(
  transformers=[
      ('num', numeric_transformer, numerical_features),
])
```

```python
# aplicar os pré-processadores
X_train = preprocessor.fit_transform(X_train)
X_test = preprocessor.transform(X_test)
X_val = preprocessor.transform(X_val)
```

```python
# Salvar preprocessor
joblib.dump(preprocessor, 'model/preprocessor.pkl')
```

```python
# Criar uma classe do tipo Dataset para os dados
class ParkisonDataset(Dataset):
  def __init__(self, X, y):
    self.X = torch.tensor(X, dtype=torch.float32)
    self.y = torch.tensor(y, dtype=torch.float32).reshape(-1, 1)

  def __len__(self):
    return len(self.X)

  def __getitem__(self, idx):
    return self.X[idx], self.y[idx]
```

```python
# Criar os datasets
dataset_train = ParkisonDataset(X_train, y_train)
dataset_val = ParkisonDataset(X_val, y_val)
dataset_test = ParkisonDataset(X_test, y_test)
```

```python
# Criar os dataloaders
dataloader_train = DataLoader(dataset_train, batch_size=1, drop_last=True)
dataloader_val = DataLoader(dataset_val, batch_size=1, drop_last=True)
dataloader_test = DataLoader(dataset_test, batch_size=1, drop_last=True)
```

## Arquitetura da Rede

```python
class NeuralNetwork(nn.Module):
    def __init__(self, input_size, hidden_layer_sizes=[128, 64, 32, 16], output_size=1, dropout_rate=0.1):
        super(NeuralNetwork, self).__init__()
        
        layers = []
        in_features = input_size

        # Iteramos sobre a lista de tamanhos. Se a lista tiver 2 itens, cria 2 camadas.
        # Se tiver 10 itens, cria 10 camadas.
        for hidden_dim in hidden_layer_sizes:
            layers.append(nn.Linear(in_features, hidden_dim))
            layers.append(nn.LayerNorm(hidden_dim))
            layers.append(nn.ReLU())
            
            # 4. Dropout
            layers.append(nn.Dropout(dropout_rate))
            
            # Atualiza a entrada da próxima camada para ser a saída da atual
            in_features = hidden_dim

        # Adicionada fora do loop, conectando a última camada oculta ao output final
        layers.append(nn.Linear(in_features, output_size))
        
        # Empacota tudo em uma sequência executável
        self.network = nn.Sequential(*layers)

    def forward(self, x):
        # O nn.Sequential executa todas as camadas na ordem que foram adicionadas
        return self.network(x)
```

## Treinar a Rede Neural

```python
def train_engine(model, train_loader, val_loader, criterion, optimizer, device, run_name, num_epochs=100, patience=35):
    """
    Executa o loop de treinamento e validação com Early Stopping e integração MLflow.
    """
    
    min_delta = 1e-6
    best_val_loss = float('inf')
    epochs_no_improvement = 0
    best_model_path = ""
    
    # Histórico para retorno (opcional)
    history = {'train_loss': [], 'val_loss': []}

    print(f"Iniciando treinamento para: {run_name}")

    for epoch in range(num_epochs):
        # --- FASE DE TREINO ---
        model.train()
        running_train_loss = 0.0
        
        for inputs, labels in train_loader:
            inputs, labels = inputs.to(device), labels.to(device)

            # Zerar gradientes
            optimizer.zero_grad()

            # Forward
            outputs = model(inputs)
            
            # Loss Calculation (BCEWithLogitsLoss espera logits)
            loss = criterion(outputs, labels)

            # Backward
            loss.backward()

            # Gradient Clipping (evita explosão do gradiente)
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

            # Atualizar Pesos
            optimizer.step()

            running_train_loss += loss.item()

        epoch_train_loss = running_train_loss / len(train_loader)
        history['train_loss'].append(epoch_train_loss)
        mlflow.log_metric("train_loss", epoch_train_loss, step=epoch)

        # --- FASE DE VALIDAÇÃO ---
        model.eval()
        running_val_loss = 0.0
        
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                
                running_val_loss += loss.item()

        epoch_val_loss = running_val_loss / len(val_loader)
        history['val_loss'].append(epoch_val_loss)
        mlflow.log_metric("val_loss", epoch_val_loss, step=epoch)

        # --- LOGS NO CONSOLE (A cada 10 épocas) ---
        if epoch % 10 == 0:
            print(f"[{run_name}] Epoch {epoch:03d} | Train Loss: {epoch_train_loss:.4f} | Val Loss: {epoch_val_loss:.4f}")

        # --- EARLY STOPPING & CHECKPOINT ---
        if (best_val_loss - epoch_val_loss) > min_delta:
            best_val_loss = epoch_val_loss
            epochs_no_improvement = 0

            best_model_path = f"model/{run_name}_best.pth"
            torch.save(model.state_dict(), best_model_path)
            
        else:
            epochs_no_improvement += 1

        if epochs_no_improvement >= patience:
            print(f"   --> Early stopping at epoch {epoch}. Best Val Loss: {best_val_loss:.4f}")
            break
    
    # Logar o artefato final no MLflow
    if best_model_path:
        mlflow.log_artifact(best_model_path)

    return best_val_loss, best_model_path
```

```python
# --- Configurações ---
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
current_experiment_name = "Parkinson_Optimization_V2"
mlflow.set_experiment(current_experiment_name)

# Definição das arquiteturas para testar
model_configs = [
    {"name": "Shallow", "layers": [32, 16]},
    {"name": "Medium",  "layers": [128, 64, 32]},
    {"name": "Deep",    "layers": [256, 128, 64, 32, 16]}
]

# Variáveis para rastrear o "Campeão dos Campeões"
global_best_loss = float('inf')
global_best_arch = []
global_best_path = ""

# Loop de Experimentos
for config in model_configs:
    run_name = config["name"]
    layers = config["layers"]
    
    # 1. Instanciar Modelo e Otimizador
    model = NeuralNetwork(input_size=X_train.shape[1], hidden_layer_sizes=layers)
    model.to(device)
    
    criterion = torch.nn.BCEWithLogitsLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
    
    # 2. Iniciar MLflow Run e Chamar a Função de Treino
    with mlflow.start_run(run_name=run_name):
        # Logar parâmetros estáticos
        mlflow.log_param("architecture", str(layers))
        mlflow.log_param("optimizer", "Adam")
        
        # CHAMADA DA FUNÇÃO QUE CRIAMOS ACIMA
        best_loss, best_path = train_engine(
            model=model,
            train_loader=dataloader_train,
            val_loader=dataloader_val,
            criterion=criterion,
            optimizer=optimizer,
            device=device,
            run_name=run_name,
            num_epochs=100,
            patience=35
        )
        
        # Verificar se é o novo campeão global
        if best_loss < global_best_loss:
            print(f"   >>> NOVO RECORDE GLOBAL! {run_name} assumiu a liderança.")
            global_best_loss = best_loss
            global_best_arch = layers
            global_best_path = best_path

# --- Resumo Final ---
print("\n" + "="*40)
print(f"MELHOR MODELO ENCONTRADO: {global_best_path}")
print(f"Arquitetura: {global_best_arch}")
print(f"Loss Final: {global_best_loss:.4f}")
print("="*40)
```

### Matriz de Confusão

```python
model = NeuralNetwork(input_size=X_train.shape[1], hidden_layer_sizes=global_best_arch)
print(f"Carregando pesos de: {global_best_path}")
model.load_state_dict(torch.load(global_best_path))
model.to(device)
```

```python
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

idx_to_class = {
    0: "Saudável",
    1: "Com Parkinson"
}

model.eval()
all_probs = []
all_preds = []
all_targets = []

with torch.no_grad():
    for data in dataloader_test:
        inputs, targets = data
        
        # Enviar inputs para o mesmo dispositivo do modelo (GPU/CPU)
        inputs = inputs.to(device) 
        
        # Calcular Logits (Saída crua da rede)
        logits = model(inputs)
        
        # --- CORREÇÃO 1: Usar Sigmoid em vez de Softmax ---
        # Transforma o logit em probabilidade (0 a 1)
        probs = torch.sigmoid(logits)
        
        # --- CORREÇÃO 2: Usar Limiar > 0.5 em vez de Argmax ---
        # Se > 50% é classe 1 (Parkinson), senão é 0 (Saudável)
        preds = (probs > 0.5).float()

        # Guardar resultados (Mover para CPU e 'achatar' arrays)
        all_probs.extend(probs.cpu().numpy().flatten())
        all_preds.extend(preds.cpu().numpy().flatten())
        all_targets.extend(targets.numpy().flatten())
```

```python
# Confusion matrix and display
labels = [idx_to_class[idx] for idx in range(len(idx_to_class))]
cm = confusion_matrix(all_targets, all_preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=labels)

fig, ax = plt.subplots(figsize=(15, 6))
disp.plot(ax=ax, cmap="Blues", values_format="d")
ax.set_title("Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
# Relatório de classificação
report = classification_report(all_targets, all_preds, zero_division=0.0, target_names=[idx_to_class[i] for i in range(len(idx_to_class))])
print("Relatório de Classificação:")
print(report)
```

```python
current_experiment = mlflow.get_experiment_by_name(current_experiment_name)
runs = mlflow.search_runs(experiment_ids=[current_experiment.experiment_id])
# Pega o ID do primeiro da lista (o menor loss)
best_run_row = runs.sort_values("metrics.val_loss", ascending=True).iloc[0]
best_run_id = best_run_row.run_id
```

```python
from sklearn.metrics import accuracy_score
test_accuracy = accuracy_score(all_targets, all_preds)
print(f"Acurácia Final no Teste: {test_accuracy:.4f}")
with mlflow.start_run(run_id=best_run_id):
        # Logar acurácia
        mlflow.log_metric("test_accuracy", test_accuracy)
        
        # (Opcional) Salvar a imagem da Matriz de Confusão no MLflow
        # Como já geramos a figura 'fig' no passo anterior, podemos salvá-la
        fig.savefig("confusion_matrix_final.png")
        mlflow.log_artifact("confusion_matrix_final.png")
        
        print("Métricas e gráficos finais anexados ao melhor run com sucesso!")
```
