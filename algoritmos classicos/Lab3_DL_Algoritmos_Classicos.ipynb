# Perceptron Simples

Este notebook apresenta uma implementação simples de perceptron, para a porta AND, sem a fase de treinamento

```python
import numpy as np

# Configuração do perceptron para a porta AND
pesos = np.array([0.5, 0.5])  # Pesos sinápticos
bias = -0.7                   # Valor do bias (ajusta o limiar de ativação)
```

```python
# Função de ativação degrau
def funcao_ativacao(soma):
    return 1 if soma >= 0 else 0

# Calcula a saída do perceptron
def calcular_saida(entrada, pesos, bias):
    # Calcula a soma ponderada: w1*x1 + w2*x2 + ... + wn*xn + bias
    soma_ponderada = np.dot(entrada, pesos) + bias
    # Aplica a função de ativação
    return funcao_ativacao(soma_ponderada), soma_ponderada
```

```python
# Dados de entrada para a porta AND
entradas = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

print("=== PERCEPTRON - PORTA LÓGICA AND ===")
print(f"Pesos: {pesos}")
print(f"Bias: {bias}")
```

```python
# Testa todas as combinações de entrada
print("\nCálculos detalhados:")
for entrada in entradas:
    saida, soma_ponderada = calcular_saida(entrada, pesos, bias)

    print(f"\nEntrada: {entrada}")
    print(f"Cálculo: ({pesos[0]} × {entrada[0]}) + ({pesos[1]} × {entrada[1]}) + ({bias}) = {soma_ponderada}")
    print(f"Função de ativação: {soma_ponderada} >= 0? {'Sim' if soma_ponderada >= 0 else 'Não'} → Saída: {saida}")
```

```python
print("\n=== TABELA VERDADE COMPLETA ===")
print("Entrada | Saída")
print("---------------")
for entrada in entradas:
    saida, _ = calcular_saida(entrada, pesos, bias)
    print(f"  {entrada}   |   {saida}")
```
