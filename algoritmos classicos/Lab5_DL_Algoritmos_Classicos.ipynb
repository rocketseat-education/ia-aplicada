```python
import numpy as np
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score

# Dados de treinamento para a porta AND
# Entradas: [x1, x2]
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Saídas esperadas (target)
y = np.array([0, 0, 0, 1])
```

```python
print("=== PERCEPTRON COM SCIKIT-LEARN ===")
print("Porta lógica AND")
print("Dados de treinamento:")
for i in range(len(X)):
    print(f"Entrada: {X[i]} → Saída esperada: {y[i]}")
```

```python
# Criando e configurando o perceptron
# max_iter: número máximo de épocas
# eta0: taxa de aprendizado
# random_state: para reproducibilidade
perceptron = Perceptron(max_iter=10, eta0=0.1, random_state=42, verbose=0)

# Treinando o perceptron
perceptron.fit(X, y)

print("\n=== RESULTADOS DO TREINAMENTO ===")
print(f"Número de épocas realizadas: {perceptron.n_iter_}")
print(f"Pesos aprendidos: {perceptron.coef_}")
print(f"Bias (intercepto) aprendido: {perceptron.intercept_}")
```

```python
# Fazendo previsões
previsoes = perceptron.predict(X)

print("\n=== PREVISÕES ===")
print("Entrada | Saída Esperada | Previsão | Correto?")
print("----------------------------------------------")
for i in range(len(X)):
    correto = "✓" if previsoes[i] == y[i] else "✗"
    print(f"  {X[i]}    |       {y[i]}        |    {previsoes[i]}     |    {correto}")

# Calculando acurácia
acuracia = accuracy_score(y, previsoes)
print(f"\nAcurácia: {acuracia * 100:.2f}%")
```

```python
# Testando o perceptron com novas entradas (mesmas da tabela verdade)
print("\n=== TESTE COMPLETO ===")
test_entradas = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

for entrada in test_entradas:
    previsao = perceptron.predict([entrada])
    print(f"Entrada: {entrada} → Saída: {previsao[0]}")
```

```python
# Mostrando a função de decisão (valores antes da função degrau)
print("\n=== VALORES DA FUNÇÃO DE DECISÃO ===")
decisao_valores = perceptron.decision_function(X)
for i in range(len(X)):
    print(f"Entrada: {X[i]} → Valor decisão: {decisao_valores[i]:.15f} → Saída: {previsoes[i]}")
```
